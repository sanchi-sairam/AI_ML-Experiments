### Tokenization using HuggingFace pre-trained models

```python
from transformers import AutoTokenizer

# Step 1: Load a pre-trained tokenizer (e.g., BERT base uncased)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Step 2: Input text to tokenize
text = "Hi Siri, I'm Sairam How are you"

# Step 3: Tokenize the text to get tokens (subword units)
tokens = tokenizer.tokenize(text)
print("Tokens:", tokens)


# Step 4: Convert tokens to token IDs (integers)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("Token IDs:", token_ids)

# Step 5: Encode the text directly to get input IDs and attention mask for the model
encoded = tokenizer(text, padding=True, truncation=True, return_tensors="pt")
print("Encoded input IDs:", encoded["input_ids"])
print("Attention mask:", encoded["attention_mask"])
```

```python
vocab = tokenizer.get_vocab()
# print(vocab)
import json
with open("../resources/bert-base_vocab.json", "w") as f:
    vocab = dict(sorted(vocab.items(), key=lambda item: item[1]))
    json.dump(vocab, f, indent=4)
    
```
