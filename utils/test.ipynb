```python
# Direct Plotting
import matplotlib.pyplot as plt
import numpy as np

# Example input & output vectors
x = np.array([2, 1, 3, 5])
y = np.array([1, 3, 2, 4])

plt.figure(figsize=(6,6))
plt.axhline(0, color='gray', linestyle='--')
plt.axvline(0, color='gray', linestyle='--')

# Plot input and output as arrows
plt.quiver(0, 0, x[0], x[1], angles='xy', scale_units='xy', scale=1, color='r', label="Input")
plt.quiver(0, 0, y[0], y[1], angles='xy', scale_units='xy', scale=1, color='b', label="Output")

plt.xlim(-1, 10)
plt.ylim(-1, 10)
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Suppose we have high-dimensional vectors
X = np.array([
    [1, 2, 3, 4, 5],  # input
    [2, 1, 4, 3, 6]   # output
])

# Reduce to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.scatter(X_pca[:,0], X_pca[:,1], c=['r','b'])
for i, txt in enumerate(["Input","Output"]):
    plt.annotate(txt, (X_pca[i,0], X_pca[i,1]))
plt.show()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import seaborn as sns

vectors = np.array([
    [1, 2, 3],  # input
    [2, 3, 4]   # output
])

sim = cosine_similarity(vectors)
sns.heatmap(sim, annot=True, cmap="coolwarm", xticklabels=["Input","Output"], yticklabels=["Input","Output"])
plt.show()
```
